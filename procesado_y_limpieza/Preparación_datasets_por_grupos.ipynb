```python
def clean_total(df_final):
  df_final['gender'] = df_final['gender'].map({'M': 0,'F': 1})
  df_final['disability'] = df_final['disability'].map({'N': 0,'Y': 1})
  df_final = pd.get_dummies(df_final, prefix="code_module", columns=["code_module"], drop_first=False)
  df_final = pd.get_dummies(df_final, prefix="region", columns=["region"], drop_first=False)
  df_final['highest_education'] = df_final['highest_education'].map({'No Formal quals': 0,'Lower Than A Level': 1, 'A Level or Equivalent': 2, 'HE Qualification': 3, 'Post Graduate Qualification': 4})
  df_final['age_band'] = df_final['age_band'].map({'0-35': 0,'35-55': 1, '55<=':2})
  df_final['result_stays']=df_final.final_result
  df_final['result_aprueba']=df_final.final_result
  df_final['result_stays'] = df_final['result_stays'].map({'Pass': 1,'Withdrawn': 0, 'Fail':1, 'Distinction': 0})
  df_final['result_aprueba'] = df_final['result_aprueba'].map({'Pass': 1,'Withdrawn': 2, 'Fail':0, 'Distinction': 1})
  df_final.drop(['final_result'], axis=1, inplace=True)
  return df_final
```

```python
def clean_gender(df):
  df['disability'] = df['disability'].map({'N': 0,'Y': 1})
  df = pd.get_dummies(df, prefix="code_module", columns=["code_module"], drop_first=False)
  df = pd.get_dummies(df, prefix="region", columns=["region"], drop_first=False)
  df['highest_education'] = df['highest_education'].map({'No Formal quals': 0,'Lower Than A Level': 1, 'A Level or Equivalent': 2, 'HE Qualification': 3, 'Post Graduate Qualification': 4})
  df['age_band'] = df['age_band'].map({'0-35': 0,'35-55': 1, '55<=':2})
  df['result_stays']=df.final_result
  df['result_aprueba']=df.final_result
  df['result_stays'] = df['result_stays'].map({'Pass': 1,'Withdrawn': 0, 'Fail':1, 'Distinction': 0})
  df['result_aprueba'] = df['result_aprueba'].map({'Pass': 1,'Withdrawn': 2, 'Fail':0, 'Distinction': 1})
  df.drop(['final_result'], axis=1, inplace=True)
  df.drop(['gender'], axis=1, inplace=True)
  return df
```
